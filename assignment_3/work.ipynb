### Implement the preprocessing and justify the preprocessing steps

```python
# --- 1. Import required libraries ---
import re
import string
import pandas as pd
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

# Download NLTK resources if you haven't already
import nltk
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('punkt_tab')
```

```python
df = pd.read_csv('cyberbullying_datasets/course_dataset/CyberBullying Comments Dataset.csv')


# --- Quick look at the dataset ---
print("✅ Data successfully loaded!")
print(f"Number of rows: {len(df)}")
print("\nFirst few rows:")
print(df.head())
```

```python
# Initialize reusable objects
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

# Function to expand simple contractions
def expand_contractions(text):
    contractions = {
        "can't": "cannot", "won't": "will not", "n't": " not",
        "'re": " are", "'s": " is", "'d": " would", "'ll": " will",
        "'t": " not", "'ve": " have", "'m": " am"
    }
    for c, expanded in contractions.items():
        text = re.sub(c, expanded, text)
    return text

# Function to clean text
def clean_text(text):
    # 1. Lowercase
    text = text.lower()
    
    # 2. Replace URLs and mentions
    text = re.sub(r'http\S+|www\S+|https\S+', ' URL ', text)
    text = re.sub(r'@\w+', ' USER ', text)
    text = re.sub(r'#', '', text)  # remove hashtag symbol, keep word
    
    # 3. Expand contractions
    text = expand_contractions(text)
    
    # 4. Remove numbers and punctuation (keep ! and ? optionally)
    text = re.sub(r'[0-9]+', '', text)
    text = text.translate(str.maketrans('', '', string.punctuation.replace('!', '').replace('?', '')))
    
    # 5. Normalize repeated characters (e.g., sooooo → sooo)
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)
    
    # 6. Remove extra whitespace
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text

# Tokenization, stopword removal, and lemmatization
def tokenize_and_lemmatize(text):
    tokens = word_tokenize(text)
    tokens = [t for t in tokens if t not in stop_words]  # optional
    tokens = [lemmatizer.lemmatize(t) for t in tokens]
    return ' '.join(tokens)

def preprocess_text(text):
    text = clean_text(text)
    text = tokenize_and_lemmatize(text)
    return text
```

```python
# Apply preprocessing
df['clean_text'] = df['Text'].apply(preprocess_text)

print(df[['Text', 'clean_text', 'CB_Label']])
```
