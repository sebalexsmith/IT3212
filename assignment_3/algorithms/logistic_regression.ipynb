```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
```

```python
df = pd.read_csv('../cyberbullying_datasets/course_dataset/cb_pca_compressed.csv.gz')
df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.drop(['bullying'], axis=1), df['bullying'], 
                                                    test_size=.3, random_state=0, stratify=df['bullying'])
```

```python
grid = {
    'C': [0.01, 0.1, 1, 10],
    'penalty': ['l1', 'l2'],
    'solver': ['lbfgs', 'liblinear', 'saga'],
}

grid_search = GridSearchCV(LogisticRegression(max_iter=10_000), grid, cv=5, scoring='accuracy').fit(X_train, y_train)
```

```python
grid_search.best_params_
```

```python
lr = LogisticRegression(max_iter=10_000, solver='liblinear').fit(X_train, y_train)
```

```python
lr.score(X_test, y_test)
```

```python
print(classification_report(y_test, lr.predict(X_test)))
```
