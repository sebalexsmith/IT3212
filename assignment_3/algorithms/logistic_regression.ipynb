```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
```

```python
df = pd.read_csv('../cyberbullying_datasets/course_dataset/cb_chi_compressed.csv.gz')
df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.drop(['bullying'], axis=1), df['bullying'], 
                                                    test_size=.2, random_state=0, stratify=df['bullying'])
```

```python
grid = {
    'C': [1, 5, 10, 15],
    'penalty': ['l2', 'l1', 'elasticnet'],
    'solver': ['lbfgs', 'liblinear', 'saga'],
    'l1_ratio': [0.2, 0.5, 0.8]
}

grid_search = GridSearchCV(LogisticRegression(random_state=0), grid, cv=3, scoring='accuracy', verbose=2).fit(X_train, y_train)
```

```python
lr = LogisticRegression(
    C=15, 
    penalty='l2', 
    solver='lbfgs', 
    random_state=0
).fit(X_train, y_train)
print(lr.score(X_test, y_test))
print(lr.score(X_train, y_train))
```

```python
print(classification_report(y_test, lr.predict(X_test)))
```
