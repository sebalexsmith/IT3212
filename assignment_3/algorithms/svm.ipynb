```python
import pandas as pd
## Chi-kvadrat data
df = pd.read_csv('../cyberbullying_datasets/course_dataset/cb_chi_compressed.csv.gz')
print(df.shape)
df
```

```python
X = df.drop(["bullying"], axis=1)
y = df["bullying"]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import classification_report, accuracy_score


X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0, stratify=y
)

from sklearn.svm import SVC

svm_model = SVC(kernel='rbf', C=1.0, gamma='scale', random_state=0)

svm_model.fit(X_train, y_train)
y_pred = svm_model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
## Her har jeg testet ulike kernels! Fant ut at RBF er best!:)

kernels = ['linear', 'poly', 'rbf']
for k in kernels:
    model = SVC(kernel=k, C=1.0, gamma='scale', random_state=0)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    print(f"Kernel: {k} → Accuracy: {acc:.4f}")
    print("Accuracy:", accuracy_score(y_test, y_pred))
    print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
#Prøvde også grid search men det ga lavere accuracy en det over, så trenger ikke å se å nøye på dette her!

from sklearn.model_selection import GridSearchCV

params = {'C': [1, 10, 20], 'gamma': ['scale', 0.001, 0.01]}
grid = GridSearchCV(SVC(kernel='rbf'), param_grid=params, cv=5)
grid.fit(X_train, y_train)
print(grid.best_params_, grid.best_score_)
```
